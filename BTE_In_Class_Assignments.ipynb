```python
weights = [10.0, 9.9, 10.1, 10.05]

s = 0

for w in weights:
  s += w

print(s/len(weights))

largest = 0
for w in weights:
  if w > largest:
    largest = w

print(largest)
```

```python
def checkEmailAddress():
  email = input("Enter an email address: ")
  assert type(email) == str

  try:
    output = '0' in email and '-' in email
    if output:
      return True
  except:
    return False
```

```python
class Greeting:
    def __init__(self, first, last):
        self.first = first
        self.last = last

    def display(self, formal=False):
        if formal:
            print(f"{self.last}, {self.first}")
        else:
            print(self.first)

    def __str__(self):
        return "Instance of class greeting"

name = Greeting("John", "Doe")
print(name)
name.display()
name.display(formal=True)
```
